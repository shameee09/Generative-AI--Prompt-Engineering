```python
### Role prompt ###

import os
from dotenv import load_dotenv
from google import genai

load_dotenv(".env", override=True)

client = genai.Client(api_key=os.getenv("GEMINI_API_KEY"))

MODEL = "gemini-3.5-flash"

question = input("Enter User Question: ")

prompt = f"""
You are an expert Python developer.

Answer the user's question as a professional Python developer.
Give a clear and simple explanation with an example when useful.

User Question:
{question}
"""

response = client.models.generate_content(
    model=MODEL,
    contents=prompt
)

print("\nResponse:")
print(response.text)
```

```python
### Few shot prompt ######

import os
from dotenv import load_dotenv
from google import genai

load_dotenv(".env", override=True)

client = genai.Client(
    api_key=os.getenv("GEMINI_API_KEY")
)

MODEL = "gemini-3.5-flash-lite"

question = input("Enter User Question: ")

prompt = f"""
You are a Python programming teacher.

Here are some examples of how to answer questions:

Example 1:
Question: What is a Python list?
Answer: A list is a collection used to store multiple values in a single variable. It is ordered and mutable.

Example 2:
Question: What is a Python tuple?
Answer: A tuple is an ordered collection of values that cannot be changed after creation.

Now answer the following question using the same style.

User Question:
{question}

Requirements:
- Give a simple definition.
- Explain briefly.
- Give one small example.
"""

response = client.models.generate_content(
    model=MODEL,
    contents=prompt
)

print("\nResponse:")
print(response.text)
```

```python
## Chaining prompt

import os
from dotenv import load_dotenv
from google import genai

load_dotenv(".env", override=True)

client = genai.Client(
    api_key=os.getenv("GEMINI_API_KEY")
)

MODEL = "gemini-3.5-flash-lite"

question = input("Enter User Question: ")

# Chain 1: Explanation
prompt1 = f"""
Explain the following topic in simple terms:

{question}
"""

response1 = client.models.generate_content(
    model=MODEL,
    contents=prompt1
)

explanation = response1.text

# Chain 2: Example
prompt2 = f"""
Based on the explanation below, provide one simple practical example.

Explanation:
{explanation}
"""

response2 = client.models.generate_content(
    model=MODEL,
    contents=prompt2
)

example = response2.text

# Chain 3: Summary
prompt3 = f"""
Create a short summary using the following explanation and example.

Explanation:
{explanation}

Example:
{example}
"""

response3 = client.models.generate_content(
    model=MODEL,
    contents=prompt3
)

print("\nFinal Response:")
print(response3.text)
```
